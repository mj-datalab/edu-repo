# Clustering text documents using k-means - solution

```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.pipeline import Pipeline
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import silhouette_score
from scipy.stats import randint
import numpy as np

# -------------------------------------
# 1) Load only the requested categories
# -------------------------------------
categories = [
    "alt.atheism",
    "talk.religion.misc",
    "comp.graphics",
    "sci.space",
]

data = fetch_20newsgroups(
    subset="all",
    categories=categories,
    remove=("headers", "footers", "quotes")
)
X_text = data.data

# -------------------------------------
# 2) Pipeline: TF-IDF ("tfidf") → K-Means ("clusterer")
# -------------------------------------
pipe = Pipeline([
    ("tfidf", TfidfVectorizer(stop_words="english", max_df=0.5)),
    ("clusterer", KMeans(
        n_clusters=4,
        init="k-means++",
        n_init="auto",
        random_state=42
    )),
])

# -------------------------------------
# 3) Silhouette scorer
# -------------------------------------
def silhouette_scorer(estimator, X):
    X_tfidf = estimator.named_steps["tfidf"].transform(X)
    labels = estimator.named_steps["clusterer"].predict(X_tfidf)

    if len(set(labels)) < 2:
        return -1.0

    return silhouette_score(X_tfidf, labels, metric="cosine")

# -------------------------------------
# 4) Search space
# -------------------------------------
param_dist = {
    "tfidf__max_df": [0.4, 0.5, 0.6, 0.7],
    "tfidf__min_df": [1, 2, 3],
    "tfidf__ngram_range": [(1,1), (1,2)],
    "clusterer__n_clusters": randint(2, 10),
}

# -------------------------------------
# 5) Randomized Search
# -------------------------------------
search = RandomizedSearchCV(
    estimator=pipe,
    param_distributions=param_dist,
    n_iter=15,
    scoring=silhouette_scorer,
    cv=3,
    verbose=1,
    random_state=42,
    n_jobs=-1
)

search.fit(X_text)

print("\nBest CV silhouette:", search.best_score_)
print("Best parameters:")
for k, v in search.best_params_.items():
    print(f"  {k}: {v}")

best_model = search.best_estimator_

# -------------------------------------
# 6) Compute full-data silhouette
# -------------------------------------
X_tfidf_full = best_model.named_steps["tfidf"].transform(X_text)
labels_full = best_model.named_steps["clusterer"].predict(X_tfidf_full)
sil_full = silhouette_score(X_tfidf_full, labels_full, metric="cosine")

print(f"\nSilhouette on full data: {sil_full:.4f}")

# -------------------------------------
# 7) Top words per cluster
# -------------------------------------
tfidf = best_model.named_steps["tfidf"]
kmeans = best_model.named_steps["clusterer"]

feature_names = np.array(tfidf.get_feature_names_out())
centroids = kmeans.cluster_centers_

n_top = 15  # number of top words to show per cluster

print("\n==============================")
print(" Top words per cluster")
print("==============================")

for cluster_idx in range(kmeans.n_clusters):
    centroid = centroids[cluster_idx]
    top_idx = centroid.argsort()[::-1][:n_top]
    top_terms = feature_names[top_idx]
    print(f"\nCluster {cluster_idx}:")
    print(", ".join(top_terms))
```
